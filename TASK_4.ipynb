```python
def validate_recharge(amount):
    valid_plans = [199, 299, 399, 599]
    if amount >= 50 and amount in valid_plans:
        return "Recharge Successful"
    return "Invalid Recharge"

while True:
    amount = int(input("Enter recharge amount: "))
    result = validate_recharge(amount)
    print(result)
    if result == "Recharge Successful":
        break
    print("Please try again.\n")
```

```python
def check_inventory(stock_dict):
    for product, quantity in stock_dict.items():
        if quantity < 15:
            print(f"{product}: Reorder Alert")
        else:
            print(f"{product}: Stock OK")

inventory = {
    "Laptop": 10,
    "Mouse": 25,
    "Keyboard": 12,
    "Monitor": 18
}

check_inventory(inventory)
```

```python
def process_result(marks):
    total = 0
    for mark in marks:
        total += mark
    average = total / len(marks)
    if average >= 50:
        return "Pass"
    return "Fail"

student_marks = [60, 45, 70, 55]
print(process_result(student_marks))
```

```python
def calculate_fare(distance, peak):
    fare = 50 + (12 * distance)
    if peak:
        fare *= 1.25
    return fare

while True:
    distance = float(input("Enter distance in km: "))
    peak_input = input("Peak hour? (yes/no): ").lower()
    peak = True if peak_input == "yes" else False

    fare = calculate_fare(distance, peak)
    print(f"Total Fare: ₹{fare:.2f}")

    retry = input("Do you want to calculate again? (yes/no): ").lower()
    if retry != "yes":
        break
```

```python
def check_attendance(attendance_list):
    present_days = 0
    for day in attendance_list:
        if day == "P":
            present_days += 1
    percentage = (present_days / len(attendance_list)) * 100
    if percentage >= 75:
        return "Eligible"
    return "Not Eligible"

attendance = ["P", "P", "A", "P", "P", "A", "P", "P"]
print(check_attendance(attendance))
```

```python
def check_password(password):
    if len(password) < 8:
        return "Weak Password"

    has_digit = False
    has_special = False
    special_chars = "@#$"

    for char in password:
        if char.isdigit():
            has_digit = True
        if char in special_chars:
            has_special = True

    if has_digit and has_special:
        return "Strong Password"
    return "Weak Password"

pwd = input("Enter password: ")
print(check_password(pwd))
```
